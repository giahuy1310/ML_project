The [Devnagari-Script dataset](https://www.openml.org/d/40923) contains 92,000 images (32x32 pixels) of 46 characters from the Devanagari script. Our goal is to learn to _read_ these letters, i.e. learn to recognize the right letter given the image.
1. Develop an image classification model to recognize handwritten characters from the Devanagari-Script dataset containing 92,000 labeled images.
2. Understand the dataset structure, including 46 character classes and 32×32 grayscale image inputs.
3. Perform data preprocessing such as normalization, reshaping, and optional augmentation (rotation, shift, noise).
4. Conduct exploratory analysis to inspect class distribution and visualize sample characters.
5. Build baseline models such as Logistic Regression and SVM or Kernelization to establish initial performance.
6. Develop Convolutional Neural Networks (CNNs) to capture spatial features for improved recognition accuracy.
7. Train models using cross-entropy loss and validate performance with accuracy and confusion matrix analysis.
8. Compare models, analyze misclassifications, and optimize performance using techniques like dropout, batch normalization, or hyperparameter tuning.
